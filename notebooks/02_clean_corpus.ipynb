Use one English-original-language movie overview per movie ID. Keep the text corpus broad and mark unavailable financial values as missing.

```python
import pandas as pd

from corpus import CLEAN_PATH, RAW_PATH, prepare_movies

raw = pd.read_csv(RAW_PATH, low_memory=False)
movies, counts = prepare_movies(raw)
display(counts.to_frame())

assert movies["movie_id"].is_unique
assert movies["overview"].str.strip().ne("").all()
assert movies["revenue"].dropna().gt(0).all()

CLEAN_PATH.parent.mkdir(parents=True, exist_ok=True)
movies[["movie_id", "original_language", "overview", "release_date",
        "revenue", "budget", "runtime"]].to_csv(CLEAN_PATH, index=False)
```
